Todos os anos, estudantes do ensino médio nos Estados Unidos fazem o SAT, um exame padronizado que mede habilidades de leitura, matemática e escrita. São três seções — leitura, matemática e escrita — cada uma com **pontuação máxima de 800 pontos**. Essas provas são extremamente importantes para estudantes e universidades, pois têm um papel central no processo de admissão.

Analisar o desempenho das escolas é importante para diversos públicos, incluindo profissionais de políticas e educação, pesquisadores, governo e até pais que estão escolhendo em qual escola matricular seus filhos.

Você recebeu um conjunto de dados chamado `schools.csv`, com uma prévia abaixo.

```python
#Seleção apenas das escolas e medias de matematica
school_math = schools[['school_name','average_math']]
#Condição para filtrar apenas por determinada pontução
max_pont = 800*0.80
best_math_schools = school_math[school_math['average_math'] >= max_pont].sort_values(by=['average_math'],ascending=False)
best_math_schools
```

```python
top_10_schools = schools[['school_name','average_math','average_reading','average_writing']]
top_10_schools['total_SAT'] = top_10_schools['average_math'] + top_10_schools['average_reading'] + top_10_schools['average_writing']
top_10_schools = top_10_schools.sort_values(by=['total_SAT'], ascending=False)
top_10_schools = top_10_schools.head(10)
top_10_schools
```

```python
largest_std_dev = schools[['borough','average_math','average_reading','average_writing']]
largest_std_dev['total_SAT'] = schools['average_math'] + schools['average_reading'] + schools['average_writing']
# largest_std_dev['average_SAT'] = largest_std_dev['total_SAT'].mean().round(2)
# largest_std_dev['std_SAT'] = largest_std_dev['total_SAT'].std().round(2)
largest_std_dev = largest_std_dev.groupby(by='borough')['total_SAT'].agg(num_schools = 'count', average_SAT = 'mean', std_SAT = 'std').round(2)
std_max = largest_std_dev['std_SAT'].max()
largest_std_dev = largest_std_dev.reset_index()
largest_std_dev = largest_std_dev[largest_std_dev['std_SAT'] == std_max]
largest_std_dev
```
